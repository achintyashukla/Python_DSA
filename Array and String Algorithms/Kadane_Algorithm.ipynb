# Kadane's Algorithm

**Kadane's Algorithm** is an optimal dynamic programming / greedy algorithm used to solve the **Maximum Subarray Problem**:

> Given an integer array `nums`, find the contiguous subarray (containing at least one number) which has the largest sum and return its sum.

---

## Core Intuition

At every index `i`, suppose we already know the maximum subarray sum ending at index `i - 1` (call it `current_sum`). For element `nums[i]`, we only have **two choices**:

1. **Extend** the previous subarray by including `nums[i]`: `current_sum + nums[i]`
2. **Start a new subarray** beginning at `nums[i]`: `nums[i]`

We pick whichever gives a larger sum:

$$\text{current\_sum} = \max(\text{nums}[i],\; \text{current\_sum} + \text{nums}[i])$$

Then we update the global maximum seen so far:

$$\text{max\_sum} = \max(\text{max\_sum},\; \text{current\_sum})$$

**Why does this work?**
- If `current_sum` before index `i` is negative, adding it to `nums[i]` will only *decrease* the sum (`current_sum + nums[i] < nums[i]`), so it is always better to discard the negative prefix and start fresh at `nums[i]`.
- Initializing both `current_sum` and `max_sum` to `nums[0]` (instead of `0`) ensures the algorithm works correctly even when **all numbers in the array are negative**.

## Example & Step-by-Step Execution

Consider the array:

```python
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
```

### Initialization (index `0`)
- `current_sum = nums[0] = -2`
- `max_sum = nums[0] = -2`
- Active subarray = `[-2]`

### Step-by-Step Trace (`i = 1` to `8`)

| Step (`i`) | `nums[i]` | `current_sum + nums[i]` | Decision: `max(nums[i], current_sum + nums[i])` | `current_sum` | `max_sum` | Current Subarray |
| :---: | :---: | :---: | :--- | :---: | :---: | :--- |
| 0 (init) | `-2` | — | Initialize with `nums[0]` | `-2` | `-2` | `[-2]` |
| 1 | `1` | `-2 + 1 = -1` | `max(1, -1) = 1` (**Restart** at `1`) | `1` | `1` | `[1]` |
| 2 | `-3` | `1 + (-3) = -2` | `max(-3, -2) = -2` (**Extend**) | `-2` | `1` | `[1, -3]` |
| 3 | `4` | `-2 + 4 = 2` | `max(4, 2) = 4` (**Restart** at `4`) | `4` | `4` | `[4]` |
| 4 | `-1` | `4 + (-1) = 3` | `max(-1, 3) = 3` (**Extend**) | `3` | `4` | `[4, -1]` |
| 5 | `2` | `3 + 2 = 5` | `max(2, 5) = 5` (**Extend**) | `5` | `5` | `[4, -1, 2]` |
| 6 | `1` | `5 + 1 = 6` | `max(1, 6) = 6` (**Extend**) | `6` | **`6`** | **`[4, -1, 2, 1]`** |
| 7 | `-5` | `6 + (-5) = 1` | `max(-5, 1) = 1` (**Extend**) | `1` | `6` | `[4, -1, 2, 1, -5]` |
| 8 | `4` | `1 + 4 = 5` | `max(4, 5) = 5` (**Extend**) | `5` | `6` | `[4, -1, 2, 1, -5, 4]` |

### Result
- **Maximum Subarray Sum:** `6`
- **Subarray:** `[4, -1, 2, 1]` (indices `3` through `6`)

````python
def max_subarray_sum(nums: list[int]) -> int:
    """Return the maximum contiguous subarray sum using Kadane's Algorithm."""
    current_sum = max_sum = nums[0]

    for num in nums[1:]:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)

    return max_sum


def kadane_with_subarray(nums: list[int]) -> tuple[int, list[int]]:
    """Return (max_sum, subarray) and print step-by-step execution."""
    current_sum = max_sum = nums[0]
    temp_start = best_start = best_end = 0

    print(f"{'i':>2} | {'nums[i]':>7} | {'Action':<8} | {'current_sum':>11} | {'max_sum':>7} | Current Window")
    print("-" * 70)
    print(f"{0:>2} | {nums[0]:>7} | {'Init':<8} | {current_sum:>11} | {max_sum:>7} | {nums[0:1]}")

    for i in range(1, len(nums)):
        if nums[i] > current_sum + nums[i]:
            current_sum = nums[i]
            temp_start = i
            action = "Restart"
        else:
            current_sum += nums[i]
            action = "Extend"

        if current_sum > max_sum:
            max_sum = current_sum
            best_start = temp_start
            best_end = i

        print(f"{i:>2} | {nums[i]:>7} | {action:<8} | {current_sum:>11} | {max_sum:>7} | {nums[temp_start:i + 1]}")

    return max_sum, nums[best_start:best_end + 1]


# Example run
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
best_sum, best_subarray = kadane_with_subarray(nums)
print(f"\nMax Sum: {best_sum}, Subarray: {best_subarray}")

# Edge cases
assert max_subarray_sum([-5, -2, -8, -1]) == -1  # All negative
assert max_subarray_sum([7]) == 7                # Single element
assert max_subarray_sum([5, 4, -1, 7, 8]) == 23  # Whole array
````

## Time and Space Complexity

| Approach | Time Complexity | Space Complexity | Notes |
| :--- | :---: | :---: | :--- |
| Brute Force (3 nested loops) | $O(n^3)$ | $O(1)$ | Recomputes sum for every `(i, j)` pair |
| Improved Brute Force / Prefix Sums | $O(n^2)$ | $O(1)$ or $O(n)$ | Running sum for each start index `i` |
| Divide and Conquer | $O(n \log n)$ | $O(\log n)$ | Splits array in half and checks crossing sum |
| **Kadane's Algorithm** | **$O(n)$** | **$O(1)$** | **Optimal: single pass with constant extra space** |

### Why Kadane's is $O(n)$ Time and $O(1)$ Space
- **Time Complexity — $O(n)$:** We traverse the array of length `n` exactly once, performing only constant-time arithmetic and `max` comparisons at each element.
- **Space Complexity — $O(1)$:** We only maintain a fixed number of scalar variables (`current_sum`, `max_sum`, and optional index pointers), regardless of input size.

### Quick Pattern Recognition Guide
- **Sum / Gain over a contiguous range:** Standard Kadane (`cur = max(x, cur + x)`).
- **Wrap-around / Circular array:** Compute both `max_subarray` and `min_subarray` via Kadane's.
- **Sign flips (products or absolute values):** Track both `cur_max` and `cur_min`.
- **Two-array difference / character frequency difference:** Convert the problem into a `+` / `-` difference array, then apply Kadane's.

````python
# Bonus: Reference templates for common Kadane variants

def max_subarray_circular(nums: list[int]) -> int:
    """LeetCode 918: Maximum Sum Circular Subarray."""
    total = 0
    cur_max = cur_min = 0
    max_sum = min_sum = nums[0]

    for num in nums:
        cur_max = max(num, cur_max + num)
        max_sum = max(max_sum, cur_max)

        cur_min = min(num, cur_min + num)
        min_sum = min(min_sum, cur_min)

        total += num

    # If all numbers are negative, max_sum is the largest single element
    return max(max_sum, total - min_sum) if max_sum > 0 else max_sum


def max_product_subarray(nums: list[int]) -> int:
    """LeetCode 152: Maximum Product Subarray."""
    cur_max = cur_min = ans = nums[0]

    for num in nums[1:]:
        candidates = (num, cur_max * num, cur_min * num)
        cur_max = max(candidates)
        cur_min = min(candidates)
        ans = max(ans, cur_max)

    return ans


def max_absolute_sum(nums: list[int]) -> int:
    """LeetCode 1749: Maximum Absolute Sum of Any Subarray."""
    cur_max = cur_min = 0
    max_sum = min_sum = 0

    for num in nums:
        cur_max = max(0, cur_max + num)
        max_sum = max(max_sum, cur_max)

        cur_min = min(0, cur_min + num)
        min_sum = min(min_sum, cur_min)

    return max(max_sum, abs(min_sum))


# Sanity checks
assert max_subarray_circular([5, -3, 5]) == 10
assert max_subarray_circular([-3, -2, -3]) == -2
assert max_product_subarray([2, 3, -2, 4]) == 6
assert max_product_subarray([-2, 0, -1]) == 0
assert max_absolute_sum([2, -5, 1, -4, 3, -2]) == 8
print("All variant checks passed!")
````
